```python
# Pregunta 1

# Datos del problema
mu = 80      # media
sigma = 12   # desviacion estandar

print("Distribución Normal de Calificaciones")
print(f"Media: {mu}, Desviación estándar: {sigma}")
print("-" * 60)

# Pregunta 1: Calificación que separa 30% (izq) y 70% (der)
print("\n1. Percentil 30 (30% izquierda, 70% derecha)")
print("\nProcedimiento:")
print("- Queremos P(X <= x) = 0.30")
print("- Usamos Z = (X - μ) / σ")
print("- Despejamos X = μ + Z*σ")

p1 = 0.30
z1 = stats.norm.ppf(p1)
x1 = mu + z1 * sigma

print(f"\nZ para p=0.30: {z1:.4f}")
print(f"X = {mu} + ({z1:.4f}) * {sigma}")
print(f"X = {x1:.2f}")
print(f"\nRespuesta: {x1:.2f}")

# Grafica 1
x_vals = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
y_vals = stats.norm.pdf(x_vals, mu, sigma)

plt.figure(figsize=(10, 6))
plt.plot(x_vals, y_vals, 'b-', linewidth=2)
plt.fill_between(x_vals[x_vals <= x1], y_vals[x_vals <= x1], alpha=0.3, color='red')
plt.axvline(x1, color='red', linestyle='--', linewidth=2, label=f'x = {x1:.2f}')
plt.axvline(mu, color='black', linestyle=':', linewidth=1, label=f'μ = {mu}')
plt.title('Pregunta 1: Percentil 30')
plt.xlabel('Calificación')
plt.ylabel('Densidad')
plt.legend()
plt.grid(True, alpha=0.3)
plt.text(x1-8, 0.01, '30%', fontsize=12, color='red')
plt.text(x1+8, 0.01, '70%', fontsize=12, color='blue')
plt.savefig('pregunta1.png', dpi=150, bbox_inches='tight')
plt.show()
print("Gráfica guardada: pregunta1.png")

# Pregunta 2: Top 10% de la clase
print("\n" + "-" * 60)
print("2. Top 10% (calificación mínima)")
print("\nProcedimiento:")
print("- Top 10% significa percentil 90")
print("- P(X <= x) = 0.90")
print("- X = μ + Z*σ")

p2 = 0.90
z2 = stats.norm.ppf(p2)
x2 = mu + z2 * sigma

print(f"\nZ para p=0.90: {z2:.4f}")
print(f"X = {mu} + ({z2:.4f}) * {sigma}")
print(f"X = {x2:.2f}")
print(f"\nRespuesta: {x2:.2f}")

# Grafica 2
plt.figure(figsize=(10, 6))
plt.plot(x_vals, y_vals, 'b-', linewidth=2)
plt.fill_between(x_vals[x_vals >= x2], y_vals[x_vals >= x2], alpha=0.3, color='gold')
plt.axvline(x2, color='orange', linestyle='--', linewidth=2, label=f'x = {x2:.2f}')
plt.axvline(mu, color='black', linestyle=':', linewidth=1, label=f'μ = {mu}')
plt.title('Pregunta 2: Top 10%')
plt.xlabel('Calificación')
plt.ylabel('Densidad')
plt.legend()
plt.grid(True, alpha=0.3)
plt.text(x2-15, 0.01, '90%', fontsize=12, color='blue')
plt.text(x2+5, 0.015, 'Top 10%', fontsize=12, color='orange', fontweight='bold')
plt.savefig('pregunta2.png', dpi=150, bbox_inches='tight')
plt.show()
print("Gráfica guardada: pregunta2.png")

# Pregunta 3: 50% central (Q1 y Q3)
print("\n" + "-" * 60)
print("3. 50% central (entre Q1 y Q3)")
print("\nProcedimiento:")
print("- Q1 es percentil 25: P(X <= Q1) = 0.25")
print("- Q3 es percentil 75: P(X <= Q3) = 0.75")
print("- Usamos X = μ + Z*σ para cada uno")

# Q1
p_q1 = 0.25
z_q1 = stats.norm.ppf(p_q1)
q1 = mu + z_q1 * sigma

print(f"\nQ1:")
print(f"Z para p=0.25: {z_q1:.4f}")
print(f"Q1 = {mu} + ({z_q1:.4f}) * {sigma} = {q1:.2f}")

# Q3
p_q3 = 0.75
z_q3 = stats.norm.ppf(p_q3)
q3 = mu + z_q3 * sigma

print(f"\nQ3:")
print(f"Z para p=0.75: {z_q3:.4f}")
print(f"Q3 = {mu} + ({z_q3:.4f}) * {sigma} = {q3:.2f}")

print(f"\nRespuesta: El 50% central está entre {q1:.2f} y {q3:.2f}")

# Grafica 3
plt.figure(figsize=(10, 6))
plt.plot(x_vals, y_vals, 'b-', linewidth=2)
plt.fill_between(x_vals[(x_vals >= q1) & (x_vals <= q3)], 
                 y_vals[(x_vals >= q1) & (x_vals <= q3)], 
                 alpha=0.4, color='purple')
plt.axvline(q1, color='purple', linestyle='--', linewidth=2, label=f'Q1 = {q1:.2f}')
plt.axvline(q3, color='purple', linestyle='--', linewidth=2, label=f'Q3 = {q3:.2f}')
plt.axvline(mu, color='black', linestyle=':', linewidth=1, label=f'μ = {mu}')
plt.title('Pregunta 3: 50% Central')
plt.xlabel('Calificación')
plt.ylabel('Densidad')
plt.legend()
plt.grid(True, alpha=0.3)
plt.text(q1-10, 0.005, '25%', fontsize=11, color='gray')
plt.text((q1+q3)/2-3, 0.02, '50%', fontsize=12, color='purple', fontweight='bold')
plt.text(q3+5, 0.005, '25%', fontsize=11, color='gray')
plt.savefig('pregunta3.png', dpi=150, bbox_inches='tight')
plt.show()
print("Gráfica guardada: pregunta3.png")

# Resumen
print("\n" + "=" * 60)
print("RESUMEN")
print("=" * 60)
print(f"1. Percentil 30: {x1:.2f}")
print(f"2. Top 10% (mínimo): {x2:.2f}")
print(f"3. 50% central: [{q1:.2f}, {q3:.2f}]")
print("=" * 60)
```

```python
# Pregunta 2

print("="*60)
print("Cálculo de Tamaño de Muestra para Encuesta")
print("="*60)

# Datos
E = 0.02  # margen de error (2%)
confianza = 0.95  # 95% de confianza

print(f"\nDatos del problema:")
print(f"- Margen de error: ±{E*100}%")
print(f"- Nivel de confianza: {confianza*100}%")

print("\n" + "-"*60)
print("PROCEDIMIENTO")
print("-"*60)

print("\nFórmula para tamaño de muestra:")
print("\n   n = (Z² × p × (1-p)) / E²")
print("\nDonde:")
print("  Z = valor crítico (distribución normal)")
print("  p = proporción estimada")
print("  E = margen de error")

# Paso 1: Encontrar Z
alpha = 1 - confianza
z = stats.norm.ppf(1 - alpha/2)

print(f"\n1. Calcular Z para confianza {confianza*100}%:")
print(f"   Z = {z:.4f}")

# Paso 2: Usar p = 0.5 (caso más conservador)
p = 0.5

print(f"\n2. Usar p = {p} (proporción más conservadora)")
print(f"   Esto maximiza p(1-p) = {p*(1-p)}")

# Paso 3: Calcular n
n = (z**2 * p * (1-p)) / (E**2)
n_final = int(np.ceil(n))

print(f"\n3. Calcular n:")
print(f"   n = ({z:.4f}² × {p} × {1-p}) / {E}²")
print(f"   n = {n:.2f}")
print(f"   n = {n_final} (redondeado hacia arriba)")

print("\n" + "="*60)
print(f"RESPUESTA: Se necesitan {n_final} personas")
print("="*60)

# Tabla comparativa
print("\nComparación con otros escenarios:")
print("-"*60)
print(f"{'Confianza':<15} {'Margen Error':<15} {'n requerido':<15}")
print("-"*60)

# Diferentes combinaciones
casos = [
    (0.90, 0.02),
    (0.95, 0.02),
    (0.99, 0.02),
    (0.95, 0.01),
    (0.95, 0.03),
    (0.95, 0.05)
]

for conf, me in casos:
    z_temp = stats.norm.ppf(1 - (1-conf)/2)
    n_temp = int(np.ceil((z_temp**2 * p * (1-p)) / (me**2)))
    print(f"{conf*100}%{' '*10} ±{me*100}%{' '*10} {n_temp}")

# Gráficas
print("\n" + "-"*60)
print("Generando gráficas...")

# Gráfica 1: n vs margen de error
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

margenes = np.linspace(0.01, 0.10, 100)
ns = [(z**2 * p * (1-p)) / (m**2) for m in margenes]

ax1.plot(margenes*100, ns, 'b-', linewidth=2)
ax1.axvline(E*100, color='r', linestyle='--', label=f'E = {E*100}%')
ax1.axhline(n_final, color='r', linestyle='--', label=f'n = {n_final}')
ax1.scatter([E*100], [n_final], color='r', s=100, zorder=5)
ax1.set_xlabel('Margen de Error (%)')
ax1.set_ylabel('Tamaño de Muestra')
ax1.set_title('Tamaño de Muestra vs Margen de Error')
ax1.legend()
ax1.grid(True, alpha=0.3)

# Gráfica 2: n vs proporción p
props = np.linspace(0.1, 0.9, 100)
ns2 = [(z**2 * pp * (1-pp)) / (E**2) for pp in props]

ax2.plot(props, ns2, 'g-', linewidth=2)
ax2.axvline(p, color='r', linestyle='--', label=f'p = {p}')
ax2.axhline(n_final, color='r', linestyle='--', label=f'n = {n_final}')
ax2.scatter([p], [n_final], color='r', s=100, zorder=5)
ax2.set_xlabel('Proporción p')
ax2.set_ylabel('Tamaño de Muestra')
ax2.set_title('Tamaño de Muestra vs Proporción')
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('tamano_muestra.png', dpi=150, bbox_inches='tight')
plt.show()

print("Gráfica guardada: tamano_muestra.png")
```

```python
# Pregunta 3

np.random.seed(42)

print("="*70)
print("PRUEBAS DE HIPÓTESIS ESTADÍSTICAS")
print("="*70)

# ============================================================================
# SITUACIÓN 1: Salario de ingenieros (Prueba de media)
# ============================================================================

print("\n" + "="*70)
print("SITUACIÓN 1: Salario de graduados en ingeniería")
print("="*70)

# Hipótesis
print("\nH₀: μ ≤ 20,000")
print("H₁: μ > 20,000  (prueba unilateral derecha)")

# Simular datos de muestra
n1 = 50  # tamaño de muestra
mu_0 = 20000  # valor de referencia
salarios = np.random.normal(21500, 3000, n1)  # datos simulados

# Estadísticos de la muestra
x_bar = np.mean(salarios)
s = np.std(salarios, ddof=1)
se = s / np.sqrt(n1)

# Estadístico t
t_stat = (x_bar - mu_0) / se
df = n1 - 1

# Valor p (unilateral derecha)
p_value = 1 - stats.t.cdf(t_stat, df)

# Valor crítico (α = 0.05)
alpha = 0.05
t_critico = stats.t.ppf(1 - alpha, df)

print(f"\nDatos de muestra: n = {n1}")
print(f"Media muestral: x̄ = ${x_bar:.2f}")
print(f"Desv. estándar: s = ${s:.2f}")
print(f"Error estándar: SE = ${se:.2f}")
print(f"\nEstadístico t = {t_stat:.4f}")
print(f"Grados de libertad = {df}")
print(f"Valor crítico (α={alpha}): t_c = {t_critico:.4f}")
print(f"Valor p = {p_value:.4f}")

if p_value < alpha:
    print(f"\nConclusión: Rechazamos H₀ (p < {alpha})")
    print("Hay evidencia de que el salario promedio > $20,000")
else:
    print(f"\nConclusión: No rechazamos H₀ (p ≥ {alpha})")
    print("No hay suficiente evidencia")

# Gráfica
fig, ax = plt.subplots(figsize=(10, 6))
x = np.linspace(-4, 4, 1000)
y = stats.t.pdf(x, df)
ax.plot(x, y, 'b-', linewidth=2, label='Distribución t')
ax.fill_between(x[x >= t_critico], y[x >= t_critico], alpha=0.3, color='red', 
                label=f'Región de rechazo (α={alpha})')
ax.axvline(t_stat, color='green', linestyle='--', linewidth=2, 
           label=f't = {t_stat:.2f}')
ax.axvline(t_critico, color='red', linestyle='--', linewidth=1)
ax.set_xlabel('Estadístico t')
ax.set_ylabel('Densidad')
ax.set_title('Situación 1: Prueba de hipótesis para la media (μ)')
ax.legend()
ax.grid(True, alpha=0.3)
plt.savefig('situacion1_salario.png', dpi=150, bbox_inches='tight')
plt.show()

# ============================================================================
# SITUACIÓN 2: Temperatura corporal (Prueba de varianza)
# ============================================================================

print("\n" + "="*70)
print("SITUACIÓN 2: Temperatura corporal")
print("="*70)

# Hipótesis
print("\nH₀: σ = 0.62")
print("H₁: σ ≠ 0.62  (prueba bilateral)")

# Simular datos
n2 = 40
sigma_0 = 0.62  # valor de referencia
temperaturas = np.random.normal(98.6, 0.65, n2)

# Estadísticos
s2 = np.std(temperaturas, ddof=1)
s2_squared = s2**2
sigma_0_squared = sigma_0**2

# Estadístico chi-cuadrado
chi2_stat = (n2 - 1) * s2_squared / sigma_0_squared
df2 = n2 - 1

# Valores críticos (bilateral, α = 0.05)
alpha2 = 0.05
chi2_lower = stats.chi2.ppf(alpha2/2, df2)
chi2_upper = stats.chi2.ppf(1 - alpha2/2, df2)

# Valor p (bilateral)
if chi2_stat < df2:
    p_value2 = 2 * stats.chi2.cdf(chi2_stat, df2)
else:
    p_value2 = 2 * (1 - stats.chi2.cdf(chi2_stat, df2))

print(f"\nDatos de muestra: n = {n2}")
print(f"Desv. estándar muestral: s = {s2:.4f}")
print(f"Estadístico χ² = {chi2_stat:.4f}")
print(f"Grados de libertad = {df2}")
print(f"Valores críticos (α={alpha2}): [{chi2_lower:.2f}, {chi2_upper:.2f}]")
print(f"Valor p = {p_value2:.4f}")

if p_value2 < alpha2:
    print(f"\nConclusión: Rechazamos H₀ (p < {alpha2})")
    print(f"La desviación estándar es diferente de {sigma_0}")
else:
    print(f"\nConclusión: No rechazamos H₀ (p ≥ {alpha2})")
    print(f"No hay evidencia de que σ ≠ {sigma_0}")

# Gráfica
fig, ax = plt.subplots(figsize=(10, 6))
x2 = np.linspace(0, 80, 1000)
y2 = stats.chi2.pdf(x2, df2)
ax.plot(x2, y2, 'b-', linewidth=2, label='Distribución χ²')
ax.fill_between(x2[x2 <= chi2_lower], y2[x2 <= chi2_lower], alpha=0.3, 
                color='red', label='Regiones de rechazo')
ax.fill_between(x2[x2 >= chi2_upper], y2[x2 >= chi2_upper], alpha=0.3, color='red')
ax.axvline(chi2_stat, color='green', linestyle='--', linewidth=2, 
           label=f'χ² = {chi2_stat:.2f}')
ax.axvline(chi2_lower, color='red', linestyle='--', linewidth=1)
ax.axvline(chi2_upper, color='red', linestyle='--', linewidth=1)
ax.set_xlabel('Estadístico χ²')
ax.set_ylabel('Densidad')
ax.set_title('Situación 2: Prueba de hipótesis para la varianza (σ²)')
ax.legend()
ax.grid(True, alpha=0.3)
plt.savefig('situacion2_temperatura.png', dpi=150, bbox_inches='tight')
plt.show()

# ============================================================================
# SITUACIÓN 3: Diabetes (Prueba de proporción)
# ============================================================================

print("\n" + "="*70)
print("SITUACIÓN 3: Diabetes en México")
print("="*70)

# Hipótesis
print("\nH₀: p ≥ 0.09")
print("H₁: p < 0.09  (prueba unilateral izquierda)")

# Simular datos
n3 = 500
p_0 = 0.09  # valor de referencia
x3 = 38  # personas con diabetes (simulado)
p_hat = x3 / n3

# Error estándar
se3 = np.sqrt(p_0 * (1 - p_0) / n3)

# Estadístico z
z_stat = (p_hat - p_0) / se3

# Valor p (unilateral izquierda)
p_value3 = stats.norm.cdf(z_stat)

# Valor crítico
alpha3 = 0.05
z_critico = stats.norm.ppf(alpha3)

print(f"\nDatos de muestra: n = {n3}")
print(f"Personas con diabetes: x = {x3}")
print(f"Proporción muestral: p̂ = {p_hat:.4f} ({p_hat*100:.2f}%)")
print(f"Error estándar: SE = {se3:.4f}")
print(f"\nEstadístico z = {z_stat:.4f}")
print(f"Valor crítico (α={alpha3}): z_c = {z_critico:.4f}")
print(f"Valor p = {p_value3:.4f}")

if p_value3 < alpha3:
    print(f"\nConclusión: Rechazamos H₀ (p < {alpha3})")
    print("Hay evidencia de que la proporción < 9%")
else:
    print(f"\nConclusión: No rechazamos H₀ (p ≥ {alpha3})")
    print("No hay suficiente evidencia")

# Gráfica
fig, ax = plt.subplots(figsize=(10, 6))
x3_vals = np.linspace(-4, 4, 1000)
y3 = stats.norm.pdf(x3_vals, 0, 1)
ax.plot(x3_vals, y3, 'b-', linewidth=2, label='Distribución Normal Estándar')
ax.fill_between(x3_vals[x3_vals <= z_critico], y3[x3_vals <= z_critico], 
                alpha=0.3, color='red', label=f'Región de rechazo (α={alpha3})')
ax.axvline(z_stat, color='green', linestyle='--', linewidth=2, 
           label=f'z = {z_stat:.2f}')
ax.axvline(z_critico, color='red', linestyle='--', linewidth=1)
ax.set_xlabel('Estadístico z')
ax.set_ylabel('Densidad')
ax.set_title('Situación 3: Prueba de hipótesis para la proporción (p)')
ax.legend()
ax.grid(True, alpha=0.3)
plt.savefig('situacion3_diabetes.png', dpi=150, bbox_inches='tight')
plt.show()

# Resumen
print("\n" + "="*70)
print("RESUMEN DE RESULTADOS")
print("="*70)
print(f"\n{'Situación':<30} {'Estadístico':<15} {'Valor p':<12} {'Decisión'}")
print("-"*70)
print(f"{'1. Salario':<30} t={t_stat:.2f}{' '*9} {p_value:.4f}{' '*6} {'Rechazar' if p_value < alpha else 'No rechazar'}")
print(f"{'2. Temperatura':<30} χ²={chi2_stat:.2f}{' '*8} {p_value2:.4f}{' '*6} {'Rechazar' if p_value2 < alpha2 else 'No rechazar'}")
print(f"{'3. Diabetes':<30} z={z_stat:.2f}{' '*10} {p_value3:.4f}{' '*6} {'Rechazar' if p_value3 < alpha3 else 'No rechazar'}")
print("="*70)
```

```python
# Pregunta 4

print("="*70)
print("PRUEBA DE HIPÓTESIS: Rating de Vinos")
print("="*70)

# Datos del problema
mu = 88.44      # media poblacional
sigma = 3.03    # desviación estándar poblacional
n = 10          # tamaño de muestra
x_bar = 90.2    # media muestral
alpha = 0.05    # nivel de significancia

print("\nDatos poblacionales:")
print(f"  μ = {mu} puntos")
print(f"  σ = {sigma} puntos")
print(f"\nDatos muestrales:")
print(f"  n = {n} vinos")
print(f"  x̄ = {x_bar} puntos")
print(f"\nNivel de significancia: α = {alpha}")

print("\n" + "="*70)
print("PLANTEAMIENTO DE HIPÓTESIS")
print("="*70)

print("\nPregunta: ¿Salió esta muestra de la población de vinos?")
print("\nComo NO tenemos dirección específica (mayor o menor),")
print("usamos una PRUEBA DE DOS COLAS")
print("\nH₀: μ = 88.44  (la muestra SÍ proviene de esta población)")
print("H₁: μ ≠ 88.44  (la muestra NO proviene de esta población)")

print("\n" + "="*70)
print("PROCEDIMIENTO")
print("="*70)

# Paso 1: Calcular error estándar
se = sigma / np.sqrt(n)

print("\n1. Calcular el error estándar:")
print(f"   SE = σ/√n = {sigma}/√{n} = {se:.4f}")

# Paso 2: Calcular estadístico z
z_stat = (x_bar - mu) / se

print(f"\n2. Calcular el estadístico z:")
print(f"   z = (x̄ - μ) / SE")
print(f"   z = ({x_bar} - {mu}) / {se:.4f}")
print(f"   z = {x_bar - mu:.2f} / {se:.4f}")
print(f"   z = {z_stat:.4f}")

# Paso 3: Encontrar valores críticos (dos colas)
z_critico_der = stats.norm.ppf(1 - alpha/2)
z_critico_izq = stats.norm.ppf(alpha/2)

print(f"\n3. Valores críticos para α = {alpha} (dos colas):")
print(f"   α/2 = {alpha/2}")
print(f"   z_crítico = ±{z_critico_der:.4f}")
print(f"   Región de rechazo: z < {z_critico_izq:.4f} o z > {z_critico_der:.4f}")

# Paso 4: Calcular valor p (dos colas)
p_value = 2 * (1 - stats.norm.cdf(abs(z_stat)))

print(f"\n4. Calcular valor p (dos colas):")
print(f"   p = 2 × P(Z > |z|)")
print(f"   p = 2 × P(Z > {abs(z_stat):.4f})")
print(f"   p = {p_value:.4f}")

# Paso 5: Decisión
print("\n" + "="*70)
print("DECISIÓN Y CONCLUSIÓN")
print("="*70)

print(f"\nCriterio 1: Comparar z con valores críticos")
print(f"  z calculado = {z_stat:.4f}")
print(f"  z crítico = ±{z_critico_der:.4f}")

if abs(z_stat) > z_critico_der:
    print(f"  |{z_stat:.4f}| > {z_critico_der:.4f} → Rechazamos H₀")
    decision1 = "Rechazar H₀"
else:
    print(f"  |{z_stat:.4f}| ≤ {z_critico_der:.4f} → No rechazamos H₀")
    decision1 = "No rechazar H₀"

print(f"\nCriterio 2: Comparar valor p con α")
print(f"  valor p = {p_value:.4f}")
print(f"  α = {alpha}")

if p_value < alpha:
    print(f"  {p_value:.4f} < {alpha} → Rechazamos H₀")
    decision2 = "Rechazar H₀"
else:
    print(f"  {p_value:.4f} ≥ {alpha} → No rechazamos H₀")
    decision2 = "No rechazar H₀"

print("\n" + "-"*70)
print("CONCLUSIÓN FINAL:")
print("-"*70)

if decision1 == "Rechazar H₀":
    print("\n✗ RECHAZAMOS H₀")
    print("\nInterpretación:")
    print(f"  Con un nivel de significancia del {alpha*100}%, hay suficiente")
    print("  evidencia estadística para concluir que la muestra de vinos")
    print("  NO proviene de la población original.")
    print(f"\n  La diferencia entre x̄={x_bar} y μ={mu} es estadísticamente")
    print("  significativa.")
else:
    print("\n✓ NO RECHAZAMOS H₀")
    print("\nInterpretación:")
    print(f"  Con un nivel de significancia del {alpha*100}%, NO hay suficiente")
    print("  evidencia estadística para concluir que la muestra sea diferente")
    print("  de la población original.")
    print(f"\n  La diferencia entre x̄={x_bar} y μ={mu} puede deberse al azar")
    print("  (variación muestral).")

# Gráfica
print("\n" + "="*70)
print("GENERANDO GRÁFICA...")
print("="*70)

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 10))

# Gráfica 1: Distribución muestral de la media
x = np.linspace(mu - 4*se, mu + 4*se, 1000)
y = stats.norm.pdf(x, mu, se)

ax1.plot(x, y, 'b-', linewidth=2, label='Distribución muestral de x̄')
ax1.fill_between(x[x <= mu + z_critico_izq*se], 
                 y[x <= mu + z_critico_izq*se], 
                 alpha=0.3, color='red', label='Región de rechazo (α/2)')
ax1.fill_between(x[x >= mu + z_critico_der*se], 
                 y[x >= mu + z_critico_der*se], 
                 alpha=0.3, color='red')
ax1.axvline(mu, color='black', linestyle=':', linewidth=2, label=f'μ = {mu}')
ax1.axvline(x_bar, color='green', linestyle='--', linewidth=2, 
            label=f'x̄ = {x_bar}')
ax1.axvline(mu + z_critico_izq*se, color='red', linestyle='--', linewidth=1)
ax1.axvline(mu + z_critico_der*se, color='red', linestyle='--', linewidth=1)

ax1.set_xlabel('Rating de vinos', fontsize=11)
ax1.set_ylabel('Densidad', fontsize=11)
ax1.set_title('Distribución Muestral de la Media', fontsize=13, fontweight='bold')
ax1.legend()
ax1.grid(True, alpha=0.3)

# Anotaciones
if abs(z_stat) > z_critico_der:
    ax1.text(x_bar, ax1.get_ylim()[1]*0.5, 'x̄ en región\nde rechazo', 
             fontsize=10, color='red', ha='center', fontweight='bold',
             bbox=dict(boxstyle='round', facecolor='yellow', alpha=0.7))
else:
    ax1.text(x_bar, ax1.get_ylim()[1]*0.5, 'x̄ en región\nde aceptación', 
             fontsize=10, color='green', ha='center', fontweight='bold',
             bbox=dict(boxstyle='round', facecolor='lightgreen', alpha=0.7))

# Gráfica 2: Distribución normal estándar (z)
x2 = np.linspace(-4, 4, 1000)
y2 = stats.norm.pdf(x2, 0, 1)

ax2.plot(x2, y2, 'b-', linewidth=2, label='Distribución Normal Estándar')
ax2.fill_between(x2[x2 <= z_critico_izq], y2[x2 <= z_critico_izq], 
                 alpha=0.3, color='red', label=f'Región de rechazo')
ax2.fill_between(x2[x2 >= z_critico_der], y2[x2 >= z_critico_der], 
                 alpha=0.3, color='red')
ax2.axvline(0, color='black', linestyle=':', linewidth=2, label='μ = 0')
ax2.axvline(z_stat, color='green', linestyle='--', linewidth=2, 
            label=f'z = {z_stat:.2f}')
ax2.axvline(z_critico_izq, color='red', linestyle='--', linewidth=1, 
            label=f'z = ±{z_critico_der:.2f}')
ax2.axvline(z_critico_der, color='red', linestyle='--', linewidth=1)

ax2.set_xlabel('Estadístico z', fontsize=11)
ax2.set_ylabel('Densidad', fontsize=11)
ax2.set_title('Distribución del Estadístico z', fontsize=13, fontweight='bold')
ax2.legend()
ax2.grid(True, alpha=0.3)

# Anotaciones
ax2.text(-2.5, 0.05, f'α/2 = {alpha/2}', fontsize=10, ha='center', color='red')
ax2.text(2.5, 0.05, f'α/2 = {alpha/2}', fontsize=10, ha='center', color='red')
ax2.text(0, 0.25, f'1-α = {1-alpha}', fontsize=11, ha='center', 
         fontweight='bold', color='blue')

plt.tight_layout()
plt.savefig('prueba_vinos.png', dpi=150, bbox_inches='tight')
plt.show()
print("Gráfica guardada: prueba_vinos.png")

# Resumen
print("\n" + "="*70)
print("RESUMEN")
print("="*70)
print(f"\nTipo de prueba: DOS COLAS (bilateral)")
print(f"Hipótesis: H₀: μ = {mu} vs H₁: μ ≠ {mu}")
print(f"Estadístico: z = {z_stat:.4f}")
print(f"Valor crítico: ±{z_critico_der:.4f}")
print(f"Valor p: {p_value:.4f}")
print(f"Decisión: {decision1}")
print("="*70)
```

```python
# Pregunta 5
print("="*70)
print("PRUEBA DE HIPÓTESIS: Comparación de Proporciones")
print("="*70)

# Datos del problema
n1 = 80     # mujeres encuestadas
x1 = 24     # mujeres que NO votan por el candidato
n2 = 65     # hombres encuestados
x2 = 11     # hombres que NO votan por el candidato
alpha = 0.10  # nivel de significancia

print("\nDatos:")
print(f"Mujeres:")
print(f"  n₁ = {n1}")
print(f"  x₁ = {x1} (NO votan por el candidato)")
print(f"\nHombres:")
print(f"  n₂ = {n2}")
print(f"  x₂ = {x2} (NO votan por el candidato)")
print(f"\nNivel de significancia: α = {alpha}")

# Calcular proporciones muestrales
p1_hat = x1 / n1
p2_hat = x2 / n2

print(f"\nProporciones muestrales:")
print(f"  p̂₁ (mujeres) = {x1}/{n1} = {p1_hat:.4f} ({p1_hat*100:.2f}%)")
print(f"  p̂₂ (hombres) = {x2}/{n2} = {p2_hat:.4f} ({p2_hat*100:.2f}%)")

print("\n" + "="*70)
print("PLANTEAMIENTO DE HIPÓTESIS")
print("="*70)

print("\nPregunta: ¿Son las proporciones significativamente diferentes?")
print("\nUsamos una PRUEBA DE DOS COLAS")
print("\nH₀: p₁ = p₂  (las proporciones son iguales)")
print("H₁: p₁ ≠ p₂  (las proporciones son diferentes)")

print("\n" + "="*70)
print("PROCEDIMIENTO")
print("="*70)

# Paso 1: Calcular proporción combinada
p_pool = (x1 + x2) / (n1 + n2)

print("\n1. Calcular proporción combinada (pooled):")
print(f"   p̄ = (x₁ + x₂) / (n₁ + n₂)")
print(f"   p̄ = ({x1} + {x2}) / ({n1} + {n2})")
print(f"   p̄ = {x1 + x2} / {n1 + n2}")
print(f"   p̄ = {p_pool:.4f}")

# Paso 2: Calcular error estándar
se = np.sqrt(p_pool * (1 - p_pool) * (1/n1 + 1/n2))

print(f"\n2. Calcular error estándar:")
print(f"   SE = √[p̄(1-p̄)(1/n₁ + 1/n₂)]")
print(f"   SE = √[{p_pool:.4f} × {1-p_pool:.4f} × (1/{n1} + 1/{n2})]")
print(f"   SE = √[{p_pool:.4f} × {1-p_pool:.4f} × {1/n1 + 1/n2:.6f}]")
print(f"   SE = {se:.4f}")

# Paso 3: Calcular estadístico z
z_stat = (p1_hat - p2_hat) / se

print(f"\n3. Calcular estadístico z:")
print(f"   z = (p̂₁ - p̂₂) / SE")
print(f"   z = ({p1_hat:.4f} - {p2_hat:.4f}) / {se:.4f}")
print(f"   z = {p1_hat - p2_hat:.4f} / {se:.4f}")
print(f"   z = {z_stat:.4f}")

# Paso 4: Valores críticos (dos colas)
z_critico = stats.norm.ppf(1 - alpha/2)

print(f"\n4. Valores críticos para α = {alpha} (dos colas):")
print(f"   α/2 = {alpha/2}")
print(f"   z_crítico = ±{z_critico:.4f}")

# Paso 5: Valor p
p_value = 2 * (1 - stats.norm.cdf(abs(z_stat)))

print(f"\n5. Calcular valor p (dos colas):")
print(f"   p = 2 × P(Z > |z|)")
print(f"   p = 2 × P(Z > {abs(z_stat):.4f})")
print(f"   p = {p_value:.4f}")

# Decisión
print("\n" + "="*70)
print("DECISIÓN Y CONCLUSIÓN")
print("="*70)

print(f"\nCriterio 1: Comparar |z| con z_crítico")
print(f"  |z| = {abs(z_stat):.4f}")
print(f"  z_crítico = {z_critico:.4f}")

if abs(z_stat) > z_critico:
    print(f"  {abs(z_stat):.4f} > {z_critico:.4f} → Rechazamos H₀")
    decision = "Rechazar H₀"
else:
    print(f"  {abs(z_stat):.4f} ≤ {z_critico:.4f} → No rechazamos H₀")
    decision = "No rechazar H₀"

print(f"\nCriterio 2: Comparar valor p con α")
print(f"  valor p = {p_value:.4f}")
print(f"  α = {alpha}")

if p_value < alpha:
    print(f"  {p_value:.4f} < {alpha} → Rechazamos H₀")
else:
    print(f"  {p_value:.4f} ≥ {alpha} → No rechazamos H₀")

print("\n" + "-"*70)
print("CONCLUSIÓN FINAL:")
print("-"*70)

if decision == "Rechazar H₀":
    print("\n✗ RECHAZAMOS H₀")
    print(f"\nCon un nivel de significancia del {alpha*100}%, SÍ hay evidencia")
    print("estadística para concluir que las proporciones son diferentes.")
    print(f"\nLas mujeres ({p1_hat*100:.2f}%) y los hombres ({p2_hat*100:.2f}%)")
    print("tienen proporciones significativamente diferentes de NO votar")
    print("por el candidato.")
else:
    print("\n✓ NO RECHAZAMOS H₀")
    print(f"\nCon un nivel de significancia del {alpha*100}%, NO hay evidencia")
    print("estadística para concluir que las proporciones sean diferentes.")
    print(f"\nLa diferencia entre mujeres ({p1_hat*100:.2f}%) y hombres ({p2_hat*100:.2f}%)")
    print("puede deberse al azar (variación muestral).")

# Intervalo de confianza para la diferencia
diff = p1_hat - p2_hat
ci_lower = diff - z_critico * se
ci_upper = diff + z_critico * se

print(f"\nIntervalo de confianza {(1-alpha)*100}% para (p₁ - p₂):")
print(f"  [{ci_lower:.4f}, {ci_upper:.4f}]")
if ci_lower <= 0 <= ci_upper:
    print("  (El intervalo contiene 0, consistente con H₀)")
else:
    print("  (El intervalo NO contiene 0, evidencia contra H₀)")

# Gráficas
print("\n" + "="*70)
print("GENERANDO GRÁFICAS...")
print("="*70)

fig = plt.figure(figsize=(14, 10))

# Gráfica 1: Comparación de proporciones
ax1 = plt.subplot(2, 2, 1)
categorias = ['Mujeres', 'Hombres']
proporciones = [p1_hat * 100, p2_hat * 100]
colores = ['#FF6B9D', '#4A90E2']

bars = ax1.bar(categorias, proporciones, color=colores, alpha=0.7, edgecolor='black')
ax1.set_ylabel('Proporción que NO vota (%)', fontsize=11)
ax1.set_title('Proporciones Muestrales', fontsize=12, fontweight='bold')
ax1.grid(True, alpha=0.3, axis='y')

for i, (bar, prop) in enumerate(zip(bars, proporciones)):
    height = bar.get_height()
    ax1.text(bar.get_x() + bar.get_width()/2., height + 1,
             f'{prop:.2f}%\n({[x1, x2][i]}/{[n1, n2][i]})',
             ha='center', va='bottom', fontsize=10, fontweight='bold')

# Gráfica 2: Distribución del estadístico z
ax2 = plt.subplot(2, 2, 2)
x = np.linspace(-4, 4, 1000)
y = stats.norm.pdf(x, 0, 1)

ax2.plot(x, y, 'b-', linewidth=2, label='Normal Estándar')
ax2.fill_between(x[x <= -z_critico], y[x <= -z_critico], 
                 alpha=0.3, color='red', label=f'Región rechazo')
ax2.fill_between(x[x >= z_critico], y[x >= z_critico], 
                 alpha=0.3, color='red')
ax2.axvline(z_stat, color='green', linestyle='--', linewidth=2, 
            label=f'z = {z_stat:.2f}')
ax2.axvline(-z_critico, color='red', linestyle='--', linewidth=1)
ax2.axvline(z_critico, color='red', linestyle='--', linewidth=1)

ax2.set_xlabel('Estadístico z', fontsize=11)
ax2.set_ylabel('Densidad', fontsize=11)
ax2.set_title('Distribución del Estadístico de Prueba', fontsize=12, fontweight='bold')
ax2.legend()
ax2.grid(True, alpha=0.3)

ax2.text(-2.5, 0.05, f'α/2={alpha/2}', fontsize=9, ha='center', color='red')
ax2.text(2.5, 0.05, f'α/2={alpha/2}', fontsize=9, ha='center', color='red')

# Gráfica 3: Tabla de datos
ax3 = plt.subplot(2, 2, 3)
ax3.axis('off')

tabla_data = [
    ['', 'Mujeres', 'Hombres'],
    ['Tamaño muestra', str(n1), str(n2)],
    ['NO votan', str(x1), str(x2)],
    ['Proporción', f'{p1_hat:.4f}', f'{p2_hat:.4f}'],
    ['', '', ''],
    ['Proporción pooled', f'{p_pool:.4f}', ''],
    ['Error estándar', f'{se:.4f}', ''],
    ['Estadístico z', f'{z_stat:.4f}', ''],
    ['Valor p', f'{p_value:.4f}', ''],
    ['Decisión', decision, '']
]

tabla = ax3.table(cellText=tabla_data, cellLoc='center', loc='center',
                  colWidths=[0.35, 0.25, 0.25])
tabla.auto_set_font_size(False)
tabla.set_fontsize(10)
tabla.scale(1, 2)

for i in range(len(tabla_data)):
    if i == 0:
        for j in range(3):
            tabla[(i, j)].set_facecolor('#4A90E2')
            tabla[(i, j)].set_text_props(weight='bold', color='white')
    elif i == len(tabla_data) - 1:
        tabla[(i, 0)].set_facecolor('#90EE90' if decision == "No rechazar H₀" else '#FFB6C1')
        tabla[(i, 0)].set_text_props(weight='bold')

ax3.set_title('Resumen de Resultados', fontsize=12, fontweight='bold', pad=20)

# Gráfica 4: Intervalo de confianza
ax4 = plt.subplot(2, 2, 4)
ax4.errorbar(0, diff, yerr=z_critico*se, fmt='o', markersize=10, 
             color='blue', capsize=10, capthick=2, linewidth=2,
             label=f'IC {(1-alpha)*100}%')
ax4.axhline(0, color='red', linestyle='--', linewidth=2, label='H₀: p₁ - p₂ = 0')
ax4.set_xlim(-0.5, 0.5)
ax4.set_ylabel('Diferencia de proporciones (p₁ - p₂)', fontsize=11)
ax4.set_title(f'Intervalo de Confianza {(1-alpha)*100}%', fontsize=12, fontweight='bold')
ax4.legend()
ax4.grid(True, alpha=0.3)
ax4.set_xticks([])

ax4.text(0.05, diff, f'{diff:.4f}', fontsize=10, fontweight='bold')
ax4.text(0.05, ci_upper, f'{ci_upper:.4f}', fontsize=9, color='blue')
ax4.text(0.05, ci_lower, f'{ci_lower:.4f}', fontsize=9, color='blue')

plt.tight_layout()
plt.savefig('prueba_proporciones.png', dpi=150, bbox_inches='tight')
plt.show()
print("Gráfica guardada: prueba_proporciones.png")

# Resumen final
print("\n" + "="*70)
print("RESUMEN")
print("="*70)
print(f"\nTipo de prueba: DOS COLAS (bilateral)")
print(f"Hipótesis: H₀: p₁ = p₂ vs H₁: p₁ ≠ p₂")
print(f"Nivel de significancia: α = {alpha}")
print(f"\nDiferencia de proporciones: {diff:.4f} ({diff*100:.2f}%)")
print(f"Estadístico z: {z_stat:.4f}")
print(f"Valor p: {p_value:.4f}")
print(f"Valores críticos: ±{z_critico:.4f}")
print(f"\nDecisión: {decision}")
print("="*70)
```

```python
# Pregunta 6

print("="*70)
print("PRUEBA T PAREADA: Calificaciones Antes y Después del Curso")
print("="*70)

# Datos: calificaciones antes y después
antes = np.array([18, 21, 16, 22, 19, 24, 17, 21, 23, 18, 
                  14, 16, 16, 19, 18, 20, 12, 22, 15, 17])
despues = np.array([22, 25, 17, 24, 16, 29, 20, 23, 19, 20, 
                    15, 15, 18, 26, 18, 24, 18, 25, 19, 16])

n = len(antes)
print(f"\nNúmero de estudiantes: n = {n}")

# Mostrar datos
print("\n" + "-"*70)
print("DATOS:")
print("-"*70)
print(f"{'Estudiante':<12} {'Antes':<10} {'Después':<10} {'Diferencia (d)'}")
print("-"*70)

for i in range(n):
    d = despues[i] - antes[i]
    signo = "+" if d > 0 else ""
    print(f"{i+1:<12} {antes[i]:<10} {despues[i]:<10} {signo}{d}")

print("\n" + "="*70)
print("INCISO 1: Calcular diferencias (d_i = después - antes)")
print("="*70)

# Calcular diferencias
diferencias = despues - antes

print(f"\nDiferencias: {diferencias}")
print(f"\nCantidad de diferencias positivas: {np.sum(diferencias > 0)}")
print(f"Cantidad de diferencias negativas: {np.sum(diferencias < 0)}")
print(f"Cantidad de diferencias cero: {np.sum(diferencias == 0)}")

print("\n" + "="*70)
print("INCISO 2: Calcular media de las diferencias (d̄)")
print("="*70)

d_bar = np.mean(diferencias)

print(f"\nd̄ = Σd_i / n")
print(f"d̄ = {np.sum(diferencias)} / {n}")
print(f"d̄ = {d_bar:.4f}")

print("\n" + "="*70)
print("INCISO 3: Calcular desviación estándar y error estándar")
print("="*70)

# Desviación estándar de las diferencias
sd = np.std(diferencias, ddof=1)

print(f"\nDesviación estándar (s_d):")
print(f"s_d = {sd:.4f}")

# Error estándar
se = sd / np.sqrt(n)

print(f"\nError estándar SE(d̄):")
print(f"SE(d̄) = s_d / √n")
print(f"SE(d̄) = {sd:.4f} / √{n}")
print(f"SE(d̄) = {se:.4f}")

print("\n" + "="*70)
print("PLANTEAMIENTO DE HIPÓTESIS")
print("="*70)

print("\nH₀: μ_d ≤ 0  (no hay mejora o empeora)")
print("H₁: μ_d > 0  (hay mejora después del curso)")
print("\nPrueba: UNILATERAL DERECHA (una cola)")

print("\n" + "="*70)
print("INCISO 4: Calcular estadístico T y valor p")
print("="*70)

# Estadístico T
t_stat = d_bar / se

print(f"\nEstadístico T:")
print(f"T = d̄ / SE(d̄)")
print(f"T = {d_bar:.4f} / {se:.4f}")
print(f"T = {t_stat:.4f}")

# Grados de libertad
df = n - 1

print(f"\nGrados de libertad: df = n - 1 = {n} - 1 = {df}")

# Valor p (una cola, derecha)
p_value = 1 - stats.t.cdf(t_stat, df)

print(f"\nValor p (una cola):")
print(f"p = P(T > {t_stat:.4f})")
print(f"p = {p_value:.4f}")

print("\n" + "="*70)
print("INCISO 5: Decisión y Conclusión")
print("="*70)

alpha = 0.05
print(f"\nNivel de significancia: α = {alpha}")

# Valor crítico
t_critico = stats.t.ppf(1 - alpha, df)
print(f"Valor crítico: t_c = {t_critico:.4f}")

print(f"\nComparación:")
print(f"  T calculado = {t_stat:.4f}")
print(f"  t crítico = {t_critico:.4f}")
print(f"  valor p = {p_value:.4f}")
print(f"  α = {alpha}")

if p_value < alpha:
    print(f"\n✓ RECHAZAMOS H₀ (p < α)")
    print(f"\nCONCLUSIÓN:")
    print(f"Con un nivel de significancia del {alpha*100}%, hay evidencia")
    print("estadística suficiente para concluir que el curso de preparación")
    print("SÍ mejoró significativamente las calificaciones de los estudiantes.")
    print(f"\nMejora promedio: {d_bar:.2f} puntos")
else:
    print(f"\n✗ NO RECHAZAMOS H₀ (p ≥ α)")
    print(f"\nCONCLUSIÓN:")
    print(f"Con un nivel de significancia del {alpha*100}%, NO hay evidencia")
    print("estadística suficiente para concluir que el curso mejoró")
    print("las calificaciones.")

# Intervalo de confianza
ic = stats.t.interval(1-alpha, df, loc=d_bar, scale=se)
print(f"\nIntervalo de confianza {(1-alpha)*100}% para μ_d:")
print(f"[{ic[0]:.4f}, {ic[1]:.4f}]")

# Gráficas
print("\n" + "="*70)
print("GENERANDO GRÁFICAS...")
print("="*70)

fig = plt.figure(figsize=(14, 10))

# Gráfica 1: Antes vs Después
ax1 = plt.subplot(2, 2, 1)
estudiantes = np.arange(1, n+1)
ax1.plot(estudiantes, antes, 'o-', label='Antes', color='red', markersize=6)
ax1.plot(estudiantes, despues, 's-', label='Después', color='green', markersize=6)
ax1.set_xlabel('Estudiante')
ax1.set_ylabel('Calificación')
ax1.set_title('Calificaciones Antes y Después del Curso', fontweight='bold')
ax1.legend()
ax1.grid(True, alpha=0.3)

# Gráfica 2: Diferencias
ax2 = plt.subplot(2, 2, 2)
colores = ['green' if d > 0 else 'red' for d in diferencias]
ax2.bar(estudiantes, diferencias, color=colores, alpha=0.7, edgecolor='black')
ax2.axhline(0, color='black', linestyle='-', linewidth=1)
ax2.axhline(d_bar, color='blue', linestyle='--', linewidth=2, label=f'd̄ = {d_bar:.2f}')
ax2.set_xlabel('Estudiante')
ax2.set_ylabel('Diferencia (Después - Antes)')
ax2.set_title('Diferencias Individuales', fontweight='bold')
ax2.legend()
ax2.grid(True, alpha=0.3, axis='y')

# Gráfica 3: Distribución t
ax3 = plt.subplot(2, 2, 3)
x = np.linspace(-4, 4, 1000)
y = stats.t.pdf(x, df)
ax3.plot(x, y, 'b-', linewidth=2, label=f'Distribución t (df={df})')
ax3.fill_between(x[x >= t_critico], y[x >= t_critico], 
                 alpha=0.3, color='red', label=f'Región rechazo (α={alpha})')
ax3.axvline(t_stat, color='green', linestyle='--', linewidth=2, 
            label=f'T = {t_stat:.2f}')
ax3.axvline(t_critico, color='red', linestyle='--', linewidth=1)
ax3.set_xlabel('Estadístico T')
ax3.set_ylabel('Densidad')
ax3.set_title('Distribución del Estadístico T', fontweight='bold')
ax3.legend()
ax3.grid(True, alpha=0.3)

# Gráfica 4: Histograma de diferencias
ax4 = plt.subplot(2, 2, 4)
ax4.hist(diferencias, bins=8, color='skyblue', edgecolor='black', alpha=0.7)
ax4.axvline(d_bar, color='red', linestyle='--', linewidth=2, label=f'd̄ = {d_bar:.2f}')
ax4.axvline(0, color='black', linestyle='-', linewidth=1, label='d = 0')
ax4.set_xlabel('Diferencia')
ax4.set_ylabel('Frecuencia')
ax4.set_title('Distribución de las Diferencias', fontweight='bold')
ax4.legend()
ax4.grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.savefig('prueba_t_pareada.png', dpi=150, bbox_inches='tight')
plt.show()
print("Gráfica guardada: prueba_t_pareada.png")

# Resumen
print("\n" + "="*70)
print("RESUMEN")
print("="*70)
print(f"\nMedia de diferencias: d̄ = {d_bar:.4f}")
print(f"Desviación estándar: s_d = {sd:.4f}")
print(f"Error estándar: SE = {se:.4f}")
print(f"Estadístico T: {t_stat:.4f}")
print(f"Grados de libertad: {df}")
print(f"Valor p: {p_value:.4f}")
print(f"Valor crítico (α={alpha}): {t_critico:.4f}")
print(f"\nDecisión: {'Rechazar H₀' if p_value < alpha else 'No rechazar H₀'}")
print("="*70)
```

```python
# Pregunta 7

print("="*70)
print("PRUEBA DE BONDAD DE AJUSTE CHI-CUADRADA")
print("Distribución de Poisson - Nacimientos")
print("="*70)

# Cargar datos
print("\nCargando datos desde: Babyboom.csv")

df = pd.read_csv("Babyboom.csv", sep='\t')
print(f"✓ Datos cargados: {len(df)} nacimientos")

# Convertir tiempo en minutos a horas
horas = df['Mins'] // 60  # Hora del día (0-23)

# Contar nacimientos por hora
nacimientos_por_hora = horas.value_counts().sort_index()

print(f"\nNacimientos por hora:")
print(nacimientos_por_hora)

# Crear histograma de frecuencias (cuántas horas tuvieron 0, 1, 2,... nacimientos)
freq_observada = nacimientos_por_hora.value_counts().sort_index()

# Completar con ceros para horas sin datos
max_nacimientos = 5
obs = np.zeros(max_nacimientos + 2)  # 0, 1, 2, 3, 4, 5, >5

for i in range(max_nacimientos + 1):
    if i in freq_observada.index:
        obs[i] = freq_observada[i]

# Agrupar >5
for i in freq_observada.index:
    if i > max_nacimientos:
        obs[-1] += freq_observada[i]

print(f"\nFrecuencias observadas (horas con x nacimientos):")
for i in range(len(obs)-1):
    print(f"  {i} nacimientos: {int(obs[i])} horas")
print(f"  >{max_nacimientos} nacimientos: {int(obs[-1])} horas")

# Datos del problema (del Cuadro 2) - frecuencias esperadas
esp = np.array([3.837242, 7.034815, 6.448464, 3.940656, 1.806101, 0.662225, 0.270497])

# Verificar que tengan la misma longitud
print(f"\nVerificación: obs tiene {len(obs)} elementos, esp tiene {len(esp)} elementos")

# Crear etiquetas
x = ['0', '1', '2', '3', '4', '5', '>5']

print("\nDatos del problema:")
print("- n = 44 bebés nacidos en 24 horas")
print("- Ajuste con distribución de Poisson")
print("- Nivel de significancia: α = 0.05")

print("\n" + "-"*70)
print("TABLA DE FRECUENCIAS")
print("-"*70)
print(f"{'x (nacimientos)':<20} {'Observada':<15} {'Esperada':<15}")
print("-"*70)

total_obs = 0
total_esp = 0
for i in range(len(x)):
    print(f"{str(x[i]):<20} {obs[i]:<15.0f} {esp[i]:<15.6f}")
    total_obs += obs[i]
    total_esp += esp[i]

print("-"*70)
print(f"{'TOTAL':<20} {total_obs:<15.0f} {total_esp:<15.6f}")

print("\n" + "="*70)
print("PLANTEAMIENTO DE HIPÓTESIS")
print("="*70)

print("\nH₀: Los datos siguen una distribución de Poisson")
print("H₁: Los datos NO siguen una distribución de Poisson")

print("\n" + "="*70)
print("PROCEDIMIENTO")
print("="*70)

print("\n1. Estadístico Chi-cuadrada:")
print("\n   χ² = Σ [(O_i - E_i)² / E_i]")
print("\n   Donde:")
print("   O_i = frecuencia observada")
print("   E_i = frecuencia esperada")

# Calcular chi-cuadrada
chi2_components = (obs - esp)**2 / esp

print("\n" + "-"*70)
print("Cálculo componente por componente:")
print("-"*70)
print(f"{'x':<10} {'O_i':<12} {'E_i':<12} {'(O-E)²/E':<15}")
print("-"*70)

chi2_stat = 0
for i in range(len(x)):
    comp = chi2_components[i]
    chi2_stat += comp
    print(f"{str(x[i]):<10} {obs[i]:<12.0f} {esp[i]:<12.4f} {comp:<15.6f}")

print("-"*70)
print(f"{'χ²':<35} {chi2_stat:<15.6f}")

print(f"\nχ² calculado = {chi2_stat:.6f}")

# Grados de libertad
k = len(x)  # número de categorías
parametros = 1  # estimamos λ de los datos
df = k - parametros - 1

print(f"\n2. Grados de libertad:")
print(f"   df = k - parámetros - 1")
print(f"   df = {k} - {parametros} - 1")
print(f"   df = {df}")

# Valor crítico
alpha = 0.05
chi2_critico = stats.chi2.ppf(1 - alpha, df)

print(f"\n3. Valor crítico (α = {alpha}):")
print(f"   χ²_crítico = {chi2_critico:.4f}")

# Valor p
p_value = 1 - stats.chi2.cdf(chi2_stat, df)

print(f"\n4. Valor p:")
print(f"   p = P(χ² > {chi2_stat:.4f})")
print(f"   p = {p_value:.4f}")

# Decisión
print("\n" + "="*70)
print("DECISIÓN Y CONCLUSIÓN")
print("="*70)

print(f"\nComparación:")
print(f"  χ² calculado = {chi2_stat:.4f}")
print(f"  χ² crítico = {chi2_critico:.4f}")
print(f"  valor p = {p_value:.4f}")
print(f"  α = {alpha}")

if p_value < alpha:
    print(f"\n✗ RECHAZAMOS H₀ (p < α)")
    print(f"\nCONCLUSIÓN:")
    print(f"Con un nivel de significancia del {alpha*100}%, hay evidencia")
    print("estadística para concluir que los datos NO siguen una")
    print("distribución de Poisson.")
    print("\nEl ajuste NO es adecuado.")
else:
    print(f"\n✓ NO RECHAZAMOS H₀ (p ≥ α)")
    print(f"\nCONCLUSIÓN:")
    print(f"Con un nivel de significancia del {alpha*100}%, NO hay evidencia")
    print("estadística suficiente para rechazar que los datos sigan una")
    print("distribución de Poisson.")
    print("\nEl ajuste ES adecuado. La distribución de Poisson es un buen")
    print("modelo para describir el número de nacimientos por hora.")

# Gráficas
print("\n" + "="*70)
print("GENERANDO GRÁFICAS...")
print("="*70)

fig = plt.figure(figsize=(14, 10))

# Gráfica 1: Comparación observado vs esperado
ax1 = plt.subplot(2, 2, 1)
x_pos = np.arange(len(x))
width = 0.35

bars1 = ax1.bar(x_pos - width/2, obs, width, label='Observado', 
                color='skyblue', edgecolor='black')
bars2 = ax1.bar(x_pos + width/2, esp, width, label='Esperado (Poisson)', 
                color='orange', edgecolor='black', alpha=0.7)

ax1.set_xlabel('Número de nacimientos por hora')
ax1.set_ylabel('Frecuencia')
ax1.set_title('Frecuencias Observadas vs Esperadas', fontweight='bold')
ax1.set_xticks(x_pos)
ax1.set_xticklabels(x)
ax1.legend()
ax1.grid(True, alpha=0.3, axis='y')

# Agregar valores encima de las barras
for bars in [bars1, bars2]:
    for bar in bars:
        height = bar.get_height()
        ax1.text(bar.get_x() + bar.get_width()/2., height,
                f'{height:.1f}', ha='center', va='bottom', fontsize=8)

# Gráfica 2: Contribuciones al chi-cuadrada
ax2 = plt.subplot(2, 2, 2)
colors = ['red' if c > chi2_critico/len(x) else 'green' for c in chi2_components]
bars = ax2.bar(x_pos, chi2_components, color=colors, alpha=0.7, edgecolor='black')

ax2.set_xlabel('Número de nacimientos por hora')
ax2.set_ylabel('Contribución a χ²')
ax2.set_title('Contribuciones Individuales al Estadístico χ²', fontweight='bold')
ax2.set_xticks(x_pos)
ax2.set_xticklabels(x)
ax2.grid(True, alpha=0.3, axis='y')

for bar, comp in zip(bars, chi2_components):
    height = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2., height,
            f'{comp:.3f}', ha='center', va='bottom', fontsize=8)

# Gráfica 3: Distribución chi-cuadrada
ax3 = plt.subplot(2, 2, 3)
x_chi = np.linspace(0, 20, 1000)
y_chi = stats.chi2.pdf(x_chi, df)

ax3.plot(x_chi, y_chi, 'b-', linewidth=2, label=f'χ² (df={df})')
ax3.fill_between(x_chi[x_chi >= chi2_critico], y_chi[x_chi >= chi2_critico], 
                 alpha=0.3, color='red', label=f'Región rechazo (α={alpha})')
ax3.axvline(chi2_stat, color='green', linestyle='--', linewidth=2, 
            label=f'χ² = {chi2_stat:.2f}')
ax3.axvline(chi2_critico, color='red', linestyle='--', linewidth=1)

ax3.set_xlabel('χ²')
ax3.set_ylabel('Densidad')
ax3.set_title('Distribución Chi-Cuadrada', fontweight='bold')
ax3.legend()
ax3.grid(True, alpha=0.3)

# Gráfica 4: Tabla resumen
ax4 = plt.subplot(2, 2, 4)
ax4.axis('off')

tabla_data = [
    ['Métrica', 'Valor'],
    ['χ² calculado', f'{chi2_stat:.4f}'],
    ['χ² crítico', f'{chi2_critico:.4f}'],
    ['Grados de libertad', str(df)],
    ['Valor p', f'{p_value:.4f}'],
    ['Nivel α', str(alpha)],
    ['', ''],
    ['Decisión', 'No rechazar H₀' if p_value >= alpha else 'Rechazar H₀'],
    ['Conclusión', 'Buen ajuste' if p_value >= alpha else 'Mal ajuste']
]

tabla = ax4.table(cellText=tabla_data, cellLoc='left', loc='center',
                  colWidths=[0.4, 0.4])
tabla.auto_set_font_size(False)
tabla.set_fontsize(10)
tabla.scale(1, 2.5)

for i in range(len(tabla_data)):
    if i == 0:
        for j in range(2):
            tabla[(i, j)].set_facecolor('#4A90E2')
            tabla[(i, j)].set_text_props(weight='bold', color='white')
    elif i >= len(tabla_data) - 2:
        tabla[(i, 0)].set_text_props(weight='bold')
        color = '#90EE90' if p_value >= alpha else '#FFB6C1'
        tabla[(i, 1)].set_facecolor(color)
        tabla[(i, 1)].set_text_props(weight='bold')

ax4.set_title('Resumen de Resultados', fontsize=12, fontweight='bold', pad=20)

plt.tight_layout()
plt.savefig('bondad_ajuste_poisson.png', dpi=150, bbox_inches='tight')
plt.show()
print("Gráfica guardada: bondad_ajuste_poisson.png")

# Resumen final
print("\n" + "="*70)
print("RESUMEN")
print("="*70)
print(f"\nPrueba: Chi-cuadrada de bondad de ajuste")
print(f"Distribución: Poisson")
print(f"χ² calculado: {chi2_stat:.4f}")
print(f"Grados de libertad: {df}")
print(f"χ² crítico (α={alpha}): {chi2_critico:.4f}")
print(f"Valor p: {p_value:.4f}")
print(f"\nDecisión: {'No rechazar H₀' if p_value >= alpha else 'Rechazar H₀'}")
print(f"Conclusión: La distribución de Poisson {'ES' if p_value >= alpha else 'NO ES'} un buen ajuste")
print("="*70)
```

```python
# Pregunta 8

print("="*70)
print("PRUEBA DE INDEPENDENCIA CHI-CUADRADA")
print("Supervivencia en el Titanic vs Tipo de Pasajero")
print("="*70)

# Datos del Cuadro 3
datos = np.array([
    [212, 202, 118, 178],  # Vivos
    [673, 123, 167, 528]   # Muertos
])

categorias_fila = ['Vivos', 'Muertos']
categorias_col = ['Tripulación', '1era Clase', '2da Clase', '3era Clase']

print("\nTabla de contingencia:")
print("-"*70)
df_tabla = pd.DataFrame(datos, 
                        index=categorias_fila,
                        columns=categorias_col)
df_tabla['Total'] = df_tabla.sum(axis=1)
totales_col = df_tabla.sum(axis=0)
df_tabla.loc['Total'] = totales_col
print(df_tabla)

print("\n" + "="*70)
print("PLANTEAMIENTO DE HIPÓTESIS")
print("="*70)

print("\nH₀: La supervivencia es INDEPENDIENTE del tipo de pasajero")
print("    (la probabilidad de sobrevivir es igual para todas las clases)")
print("\nH₁: La supervivencia NO es independiente del tipo de pasajero")
print("    (el tipo de pasajero afecta la probabilidad de sobrevivir)")

print("\n" + "="*70)
print("PROCEDIMIENTO")
print("="*70)

# Calcular totales
n = datos.sum()  # total general
totales_fila = datos.sum(axis=1)
totales_col = datos.sum(axis=0)

print(f"\n1. Totales:")
print(f"   Total general: n = {n}")
print(f"   Totales por fila (Vivos/Muertos): {totales_fila}")
print(f"   Totales por columna (Clases): {totales_col}")

# Calcular frecuencias esperadas
print(f"\n2. Frecuencias esperadas:")
print(f"   E_ij = (Total_fila_i × Total_col_j) / n")

esperadas = np.zeros_like(datos, dtype=float)
for i in range(len(categorias_fila)):
    for j in range(len(categorias_col)):
        esperadas[i, j] = (totales_fila[i] * totales_col[j]) / n

print("\nTabla de frecuencias esperadas:")
print("-"*70)
df_esperadas = pd.DataFrame(esperadas, 
                            index=categorias_fila,
                            columns=categorias_col)
print(df_esperadas.round(2))

# Calcular chi-cuadrada
print(f"\n3. Estadístico Chi-cuadrada:")
print(f"   χ² = Σ [(O_ij - E_ij)² / E_ij]")

chi2_components = (datos - esperadas)**2 / esperadas

print("\nContribuciones al estadístico χ²:")
print("-"*70)
df_contrib = pd.DataFrame(chi2_components, 
                          index=categorias_fila,
                          columns=categorias_col)
print(df_contrib.round(4))

chi2_stat = chi2_components.sum()
print(f"\nχ² calculado = {chi2_stat:.4f}")

# Grados de libertad
filas = len(categorias_fila)
columnas = len(categorias_col)
df_chi = (filas - 1) * (columnas - 1)

print(f"\n4. Grados de libertad:")
print(f"   df = (filas - 1) × (columnas - 1)")
print(f"   df = ({filas} - 1) × ({columnas} - 1)")
print(f"   df = {df_chi}")

# Nivel de significancia
alpha = 0.05
chi2_critico = stats.chi2.ppf(1 - alpha, df_chi)

print(f"\n5. Valor crítico (α = {alpha}):")
print(f"   χ²_crítico = {chi2_critico:.4f}")

# Valor p
p_value = 1 - stats.chi2.cdf(chi2_stat, df_chi)

print(f"\n6. Valor p:")
print(f"   p = P(χ² > {chi2_stat:.4f})")
print(f"   p = {p_value:.6f}")

# Decisión
print("\n" + "="*70)
print("DECISIÓN Y CONCLUSIÓN")
print("="*70)

print(f"\nComparación:")
print(f"  χ² calculado = {chi2_stat:.4f}")
print(f"  χ² crítico = {chi2_critico:.4f}")
print(f"  valor p = {p_value:.6f}")
print(f"  α = {alpha}")

if p_value < alpha:
    print(f"\n✗ RECHAZAMOS H₀ (p < α)")
    print(f"\nCONCLUSIÓN:")
    print(f"Con un nivel de significancia del {alpha*100}%, hay evidencia")
    print("estadística suficiente para concluir que la supervivencia")
    print("NO es independiente del tipo de pasajero.")
    print("\nEs decir, el tipo de pasajero (clase) SÍ afectó las")
    print("probabilidades de supervivencia en el Titanic.")
else:
    print(f"\n✓ NO RECHAZAMOS H₀ (p ≥ α)")
    print(f"\nCONCLUSIÓN:")
    print(f"Con un nivel de significancia del {alpha*100}%, NO hay evidencia")
    print("estadística para concluir que la supervivencia dependa del")
    print("tipo de pasajero.")

# Calcular proporciones de supervivencia
print("\n" + "-"*70)
print("ANÁLISIS ADICIONAL: Proporciones de supervivencia")
print("-"*70)

for j, clase in enumerate(categorias_col):
    vivos = datos[0, j]
    total_clase = totales_col[j]
    prop = vivos / total_clase
    print(f"{clase:<15}: {vivos}/{total_clase} = {prop:.2%} de supervivencia")

# Gráficas
print("\n" + "="*70)
print("GENERANDO GRÁFICAS...")
print("="*70)

fig = plt.figure(figsize=(14, 10))

# Gráfica 1: Frecuencias observadas
ax1 = plt.subplot(2, 2, 1)
x_pos = np.arange(len(categorias_col))
width = 0.35

bars1 = ax1.bar(x_pos - width/2, datos[0], width, label='Vivos', 
                color='green', alpha=0.7, edgecolor='black')
bars2 = ax1.bar(x_pos + width/2, datos[1], width, label='Muertos', 
                color='red', alpha=0.7, edgecolor='black')

ax1.set_xlabel('Tipo de Pasajero')
ax1.set_ylabel('Frecuencia')
ax1.set_title('Frecuencias Observadas', fontweight='bold')
ax1.set_xticks(x_pos)
ax1.set_xticklabels(categorias_col, rotation=15, ha='right')
ax1.legend()
ax1.grid(True, alpha=0.3, axis='y')

# Gráfica 2: Proporciones de supervivencia
ax2 = plt.subplot(2, 2, 2)
proporciones = datos[0] / totales_col
colors = ['#2ecc71', '#3498db', '#9b59b6', '#e74c3c']

bars = ax2.bar(x_pos, proporciones * 100, color=colors, alpha=0.7, edgecolor='black')
ax2.set_xlabel('Tipo de Pasajero')
ax2.set_ylabel('Porcentaje de Supervivencia (%)')
ax2.set_title('Tasa de Supervivencia por Tipo de Pasajero', fontweight='bold')
ax2.set_xticks(x_pos)
ax2.set_xticklabels(categorias_col, rotation=15, ha='right')
ax2.grid(True, alpha=0.3, axis='y')

for bar, prop in zip(bars, proporciones):
    height = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2., height + 1,
            f'{prop*100:.1f}%', ha='center', va='bottom', fontweight='bold')

# Gráfica 3: Mapa de calor de contribuciones
ax3 = plt.subplot(2, 2, 3)
im = ax3.imshow(chi2_components, cmap='YlOrRd', aspect='auto')

ax3.set_xticks(np.arange(len(categorias_col)))
ax3.set_yticks(np.arange(len(categorias_fila)))
ax3.set_xticklabels(categorias_col, rotation=15, ha='right')
ax3.set_yticklabels(categorias_fila)

for i in range(len(categorias_fila)):
    for j in range(len(categorias_col)):
        text = ax3.text(j, i, f'{chi2_components[i, j]:.2f}',
                       ha="center", va="center", color="black", fontweight='bold')

ax3.set_title('Contribuciones al Estadístico χ²', fontweight='bold')
plt.colorbar(im, ax=ax3)

# Gráfica 4: Distribución chi-cuadrada
ax4 = plt.subplot(2, 2, 4)
x_chi = np.linspace(0, chi2_stat + 10, 1000)
y_chi = stats.chi2.pdf(x_chi, df_chi)

ax4.plot(x_chi, y_chi, 'b-', linewidth=2, label=f'χ² (df={df_chi})')
ax4.fill_between(x_chi[x_chi >= chi2_critico], y_chi[x_chi >= chi2_critico], 
                 alpha=0.3, color='red', label=f'Región rechazo (α={alpha})')
ax4.axvline(chi2_stat, color='green', linestyle='--', linewidth=2, 
            label=f'χ² = {chi2_stat:.2f}')
ax4.axvline(chi2_critico, color='red', linestyle='--', linewidth=1)

ax4.set_xlabel('χ²')
ax4.set_ylabel('Densidad')
ax4.set_title('Distribución Chi-Cuadrada', fontweight='bold')
ax4.legend()
ax4.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('prueba_independencia_titanic.png', dpi=150, bbox_inches='tight')
plt.show()
print("Gráfica guardada: prueba_independencia_titanic.png")

# Resumen final
print("\n" + "="*70)
print("RESUMEN")
print("="*70)
print(f"\nPrueba: Chi-cuadrada de independencia")
print(f"Variables: Supervivencia × Tipo de Pasajero")
print(f"χ² calculado: {chi2_stat:.4f}")
print(f"Grados de libertad: {df_chi}")
print(f"χ² crítico (α={alpha}): {chi2_critico:.4f}")
print(f"Valor p: {p_value:.6f}")
print(f"\nDecisión: {'Rechazar H₀' if p_value < alpha else 'No rechazar H₀'}")
print(f"Conclusión: Las variables {'NO son' if p_value < alpha else 'son'} independientes")
print("="*70)
```
